```
from Bio import SeqIO
import numpy as np
import hashlib

# Reverse complement and canonical k-mer
def reverse_complement(seq):
    return seq.translate(str.maketrans("ACGT", "TGCA"))[::-1]

def canonical_kmer(kmer):
    rc = reverse_complement(kmer)
    return min(kmer, rc)

# k-mer vector extraction
def fasta_to_kmer_vector(fasta_path, k=6, vector_size=4096):
    vec = np.zeros(vector_size, dtype=np.float32)

    for record in SeqIO.parse(fasta_path, "fasta"):
        seq = str(record.seq).upper()
        for i in range(len(seq) - k + 1):
            kmer = seq[i:i+k]
            if "N" in kmer:
                continue
            kmer = canonical_kmer(kmer)
            idx = int(hashlib.md5(kmer.encode()).hexdigest(), 16) % vector_size
            vec[idx] += 1

    # Normalize so row sums to 1
    if vec.sum() > 0:
        vec /= vec.sum()

    return vec
```

```
import glob
from tqdm import tqdm

FASTA_FILES = glob.glob("ncbi_genomes/ncbi_dataset/data/*/*genomic.fna")

X = []
sample_ids = []

for fasta in tqdm(FASTA_FILES):
    sample_id = fasta.split("/")[-1].replace("_genomic.fna", "")
    vec = fasta_to_kmer_vector(fasta, k=6, vector_size=4096)
    X.append(vec)
    sample_ids.append(sample_id)

X = np.vstack(X)
print("Feature matrix shape:", X.shape)
```

```
import pandas as pd
import re
from pathlib import Path

# ---------- INPUTS ----------
KMER_IDS = "/content/sample_ids_k6.csv"
EXCEL = "/content/original 1000 samples.xlsx"

if not Path(KMER_IDS).exists():
    raise FileNotFoundError(f"Missing {KMER_IDS}")
if not Path(EXCEL).exists():
    raise FileNotFoundError(f"Missing {EXCEL}")

# ---------- NORMALIZER ----------
acc_re = re.compile(r"(GC[AF]_\d+\.\d+)")
def norm_id(x):
    m = acc_re.search(str(x))
    return m.group(1) if m else str(x).strip()

# ---------- LOAD 769 ASSEMBLIES (from k-mer IDs) ----------
kmer_df = pd.read_csv(KMER_IDS)
kmer_col = kmer_df.columns[0]
assemblies_769 = kmer_df[kmer_col].astype(str).map(norm_id).tolist()

print("Total k-mer assemblies:", len(assemblies_769))

# ---------- LOAD organism group from Excel ----------
df = pd.read_excel(EXCEL)

# Confirm required columns exist
if "assembly" not in df.columns or "organism group" not in df.columns:
    raise ValueError(
        "Excel must contain columns: 'assembly' and 'organism group'\n"
        f"Found columns: {list(df.columns)}"
    )

org_map = df[["assembly", "organism group"]].dropna()
org_map["assembly"] = org_map["assembly"].astype(str).map(norm_id)
org_map["organism group"] = org_map["organism group"].astype(str).str.strip()

# keep first mapping per assembly
org_map = org_map.drop_duplicates(subset=["assembly"], keep="first")

# ---------- MATCH ----------
org_dict = dict(zip(org_map["assembly"], org_map["organism group"]))

matched = []
missing = []

for a in assemblies_769:
    if a in org_dict:
        matched.append({"assembly": a, "organism_group": org_dict[a]})
    else:
        missing.append(a)

matched_df = pd.DataFrame(matched)

# Save results
matched_df.to_csv("/content/assembly_organism_group_769.csv", index=False)

print("\n✅ Saved: /content/assembly_organism_group_769.csv")
print("Matched rows:", matched_df.shape[0])
print("Missing organism group:", len(missing))

if missing:
    print("\n⚠️ Example missing (first 20):")
    print(missing[:20])
```

```
import pandas as pd
import re
import ast
from pathlib import Path

# --- INPUTS ---
KMER_IDS = "/content/sample_ids_k6.csv"                          # 769 k-mer IDs
ORG_MAP  = "/content/assembly_organism_group_769.csv"            # 769 mapping you created
AB_FILE  = "/content/assembly_level_antibiotics_from_AST.csv"    # upload this into Colab Files panel

for p in [KMER_IDS, ORG_MAP, AB_FILE]:
    if not Path(p).exists():
        raise FileNotFoundError(f"Missing {p}. Upload/copy it into /content.")

# --- Normalize assembly IDs to accession only (GCA/GCF_... .#) ---
acc_re = re.compile(r"(GC[AF]_\d+\.\d+)")
def norm_id(x):
    m = acc_re.search(str(x))
    return m.group(1) if m else str(x).strip()

# --- Load 769 k-mer assembly IDs (master order) ---
km = pd.read_csv(KMER_IDS)
km_ids_raw = km[km.columns[0]].astype(str).tolist()
km_ids = [norm_id(x) for x in km_ids_raw]
km_df = pd.DataFrame({"assembly": km_ids})

# --- Load organism group map (already 769) ---
org = pd.read_csv(ORG_MAP)
org["assembly"] = org["assembly"].astype(str).map(norm_id)

# --- Load antibiotic list file ---
ab = pd.read_csv(AB_FILE)
ab["assembly"] = ab["assembly"].astype(str).map(norm_id)

# Parse the antibiotic column if it's stored like "['a','b',...]"
def parse_list(x):
    if pd.isna(x):
        return []
    s = str(x).strip()
    try:
        v = ast.literal_eval(s)
        if isinstance(v, list):
            return [str(i).strip() for i in v]
    except:
        pass
    # fallback: split by comma
    return [t.strip() for t in s.split(",") if t.strip()]

ab["antibiotic_list"] = ab["antibiotic"].apply(parse_list)

# Keep one row per assembly (if duplicates exist)
ab_agg = ab.groupby("assembly", as_index=False).agg({
    "antibiotic_list": lambda L: sorted(set([a for sub in L for a in sub]))
})

# --- Merge: km order -> organism group -> antibiotic list ---
out = km_df.merge(org[["assembly","organism_group"]], on="assembly", how="left")
out = out.merge(ab_agg[["assembly","antibiotic_list"]], on="assembly", how="left")

# Fill missing antibiotic lists with empty list
out["antibiotic_list"] = out["antibiotic_list"].apply(lambda x: x if isinstance(x, list) else [])

# Create a single column like your example (string representation of list)
out["antibiotic"] = out["antibiotic_list"].apply(lambda L: str(L))

# Keep only the columns you want (like your example file)
out2 = out[["assembly", "organism_group", "antibiotic"]].copy()

# --- Save ---
out_path = "/content/assembly_organism_antibiotics_769.csv"
out2.to_csv(out_path, index=False)

print("✅ Saved:", out_path)
print("Rows:", out2.shape[0])
print("Missing organism_group:", out2["organism_group"].isna().sum())
print("Assemblies with 0 antibiotics:", (out["antibiotic_list"].apply(len) == 0).sum())
display(out2.head(10))
```

```
import pandas as pd
import numpy as np
import ast
from pathlib import Path

IN_PATH = "/content/assembly_organism_antibiotics_769.csv"
if not Path(IN_PATH).exists():
    raise FileNotFoundError(f"Missing {IN_PATH}")

df = pd.read_csv(IN_PATH)

# Parse the antibiotic list column (stored like "['a','b','c']")
def parse_list(x):
    if pd.isna(x):
        return []
    s = str(x).strip()
    try:
        v = ast.literal_eval(s)
        if isinstance(v, list):
            return [str(i).strip().lower() for i in v]
    except:
        pass
    # fallback: split by comma
    return [t.strip().lower() for t in s.split(",") if t.strip()]

df["antibiotic_list"] = df["antibiotic"].apply(parse_list)

# Collect all unique antibiotics
all_ab = sorted(set(a for L in df["antibiotic_list"] for a in L))
print("Total unique antibiotics found:", len(all_ab))

# Keep exactly 36 (like the study) using most frequent antibiotics
ab_freq = {}
for L in df["antibiotic_list"]:
    for a in set(L):
        ab_freq[a] = ab_freq.get(a, 0) + 1

top36 = sorted(ab_freq.keys(), key=lambda k: ab_freq[k], reverse=True)[:36]
print("Selected antibiotics (36):", len(top36))

# Build matrix
assembly_ids = df["assembly"].astype(str).to_numpy(dtype=object)

y = np.zeros((len(df), len(top36)), dtype=np.int32)

ab_index = {a:i for i,a in enumerate(top36)}
for i, L in enumerate(df["antibiotic_list"]):
    for a in set(L):
        if a in ab_index:
            y[i, ab_index[a]] = 1

# Save
np.save("/content/assembly_ids_769.npy", assembly_ids)
np.save("/content/y_antibiotic_769x36.npy", y)
pd.DataFrame(y, columns=[f"ab_{a}" for a in top36]).assign(assembly=assembly_ids)\
  .to_csv("/content/antibiotic_label_matrix_769x36.csv", index=False)

print("\n✅ Saved:")
print(" - /content/assembly_ids_769.npy", assembly_ids.shape)
print(" - /content/y_antibiotic_769x36.npy", y.shape, "unique:", np.unique(y))
print(" - /content/antibiotic_label_matrix_769x36.csv")
```
