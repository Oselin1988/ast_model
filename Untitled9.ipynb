```python
import pandas as pd
import re

# Load raw dataset
df = pd.read_excel("/content/original 1000 samples.xlsx")

# Clean column names
df.columns = df.columns.str.strip()

# Column names (exact from your file)
isolate_col = "isolaote identifier"
organism_col = "organism group"
assembly_col = "assembly"
source_col = "isolation source"
type_col = "isolation type of pathogen"
ast_col = "AST phenotype"

# Keep only rows with assemblies
df = df[df[assembly_col].notna()].copy()

print("Rows with assemblies:", df.shape[0])

# Parse antibiotic and phenotype
def parse_ast(value):
    if pd.isna(value):
        return None, None

    value = str(value)

    phenotype = None
    for p in ["Resistant", "Intermediate", "Susceptible"]:
        if p.lower() in value.lower():
            phenotype = p
            break

    antibiotic = re.sub(
        "(?i)resistant|intermediate|susceptible|[-:]", "", value
    ).strip()

    if not antibiotic or not phenotype:
        return None, None

    return antibiotic, phenotype

parsed = df[ast_col].apply(parse_ast)
df["antibiotic"] = parsed.apply(lambda x: x[0])
df["phenotype"] = parsed.apply(lambda x: x[1])

# Drop rows without valid AST
df = df.dropna(subset=["antibiotic", "phenotype"])

# Aggregate per isolate (GENOME-LEVEL)
rows = []

for isolate_id, g in df.groupby(isolate_col):
    rows.append({
        "isolate_id": isolate_id,
        "organism": g[organism_col].iloc[0],
        "assembly": g[assembly_col].iloc[0],
        "isolation_source": g[source_col].iloc[0],
        "isolation_type": g[type_col].iloc[0],
        "resistant_antibiotics": "; ".join(
            sorted(g[g["phenotype"] == "Resistant"]["antibiotic"].unique())
        ),
        "intermediate_antibiotics": "; ".join(
            sorted(g[g["phenotype"] == "Intermediate"]["antibiotic"].unique())
        ),
        "susceptible_antibiotics": "; ".join(
            sorted(g[g["phenotype"] == "Susceptible"]["antibiotic"].unique())
        ),
    })

clean_df = pd.DataFrame(rows)

# Save clean file
output_path = "/content/clean_771_genome_level_AST.csv"
clean_df.to_csv(output_path, index=False)

print("✅ Clean genome-level AST file created")
print("📁 Saved to:", output_path)
print("🧬 Total genomes:", clean_df.shape[0])
print("🧬 Unique assemblies:", clean_df['assembly'].nunique())
```

```python
from Bio import SeqIO
import os

# QC parameters
MIN_CONTIG_LEN = 500
MAX_N_PERCENT = 5.0

# Input and output directories
INPUT_DIR = "/content/genomes"
OUTPUT_DIR = "/content/genomes_qc"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Logging
qc_log = []

def n_percentage(seq):
    return (seq.upper().count("N") / len(seq)) * 100

# Process each assembly folder
for root, _, files in os.walk(INPUT_DIR):
    for file in files:
        if file.endswith(".fna"):
            input_fasta = os.path.join(root, file)
            assembly_id = file.split(".")[0]
            output_fasta = os.path.join(OUTPUT_DIR, f"{assembly_id}.fna")

            kept_contigs = []
            removed_contigs = 0

            for record in SeqIO.parse(input_fasta, "fasta"):
                seq_len = len(record.seq)
                n_pct = n_percentage(str(record.seq))

                if seq_len >= MIN_CONTIG_LEN and n_pct <= MAX_N_PERCENT:
                    kept_contigs.append(record)
                else:
                    removed_contigs += 1

            if kept_contigs:
                SeqIO.write(kept_contigs, output_fasta, "fasta")

            qc_log.append({
                "assembly": assembly_id,
                "input_contigs": len(list(SeqIO.parse(input_fasta, "fasta"))),
                "kept_contigs": len(kept_contigs),
                "removed_contigs": removed_contigs
            })

print(f"✅ Genome QC filtering complete")
print(f"📁 Cleaned genomes saved to: {OUTPUT_DIR}")
print(f"🧬 Total genomes processed: {len(qc_log)}")
```

```python
from Bio import SeqIO
import os
from collections import Counter
import pandas as pd

# =======================
# PARAMETERS
# =======================
K = 6
INPUT_DIR = "/content/genomes_qc"   # folder containing .fna genomes
OUTPUT_FILE = "/content/kmer_freq_table_k6.csv"

# =======================
# FUNCTION: Extract k-mers
# =======================
def extract_kmers(sequence, k):
    return [sequence[i:i+k] for i in range(len(sequence) - k + 1)]

# =======================
# MAIN: Process genomes
# =======================
all_genomes = []
genome_ids = []

for file in os.listdir(INPUT_DIR):
    if file.endswith(".fna"):
        genome_id = file.replace(".fna", "")
        input_fasta = os.path.join(INPUT_DIR, file)

        kmer_counts = Counter()
        total_kmers = 0

        for record in SeqIO.parse(input_fasta, "fasta"):
            seq = str(record.seq).upper()

            # Optional: remove non-ATGC characters
            seq = seq.replace("N", "")

            if len(seq) >= K:
                kmers = extract_kmers(seq, K)
                kmer_counts.update(kmers)
                total_kmers += len(kmers)

        # Normalize counts → frequency
        if total_kmers > 0:
            kmer_freq = {k: v / total_kmers for k, v in kmer_counts.items()}
        else:
            kmer_freq = {}

        all_genomes.append(kmer_freq)
        genome_ids.append(genome_id)

print(f"✅ Processed genomes: {len(genome_ids)}")

# =======================
# Convert to DataFrame
# =======================
df = pd.DataFrame(all_genomes)
df.insert(0, "genome_id", genome_ids)

# Replace missing k-mers with 0 frequency
df = df.fillna(0)

# Save output
df.to_csv(OUTPUT_FILE, index=False)

print("✅ Normalized k-mer frequency table saved!")
print(f"📁 Output file: {OUTPUT_FILE}")
print("📌 Shape:", df.shape)
```

```python
import pandas as pd
import re

labels_df = pd.read_excel("/content/cleaned_AST_assembly_organism.xlsx")

# If AST phenotype contains per-antibiotic results, parse it.
# Adjust separators/patterns if your format differs.
def parse_ast_list(ast_text):
    """
    Returns list of (antibiotic, phenotype) pairs from a string.
    Supports formats like:
      'ciprofloxacin: Resistant; tetracycline: Susceptible'
      'ciprofloxacin=R; tetracycline=S'
    """
    if pd.isna(ast_text):
        return []

    s = str(ast_text).strip()

    # split on ; or , or |  (common separators)
    parts = re.split(r"[;|,]\s*", s)

    pairs = []
    for p in parts:
        if not p.strip():
            continue

        # match "drug: phenotype" or "drug=phenotype"
        m = re.match(r"^\s*([A-Za-z0-9_\-\s\/]+)\s*[:=]\s*([A-Za-z]+)\s*$", p.strip())
        if m:
            drug = m.group(1).strip().lower()
            pheno = m.group(2).strip().upper()   # R/S/I style
            pairs.append((drug, pheno))
        else:
            # try looser: "drug phenotype"
            m2 = re.match(r"^\s*([A-Za-z0-9_\-\s\/]+)\s+(RESISTANT|SUSCEPTIBLE|INTERMEDIATE|R|S|I)\s*$", p.strip(), re.I)
            if m2:
                drug = m2.group(1).strip().lower()
                ph = m2.group(2).strip().upper()
                ph = {"RESISTANT":"R","SUSCEPTIBLE":"S","INTERMEDIATE":"I"}.get(ph, ph)
                pairs.append((drug, ph))
    return pairs

rows = []
for _, r in labels_df.iterrows():
    asm = str(r["assembly"]).strip()
    org = str(r["organism group"]).strip()
    ast = r["AST phenotype"]

    for drug, pheno in parse_ast_list(ast):
        rows.append({"assembly": asm, "organism group": org, "antibiotic": drug, "phenotype": pheno})

long_labels = pd.DataFrame(rows)

print("✅ Long labels shape:", long_labels.shape)
print(long_labels.head())

long_labels.to_csv("/content/labels_long_antibiotic.csv", index=False)
print("📁 Saved: /content/labels_long_antibiotic.csv")
```

```python
import pandas as pd
import re

df = pd.read_excel("/content/original 1000 samples.xlsx")
df.columns = df.columns.str.strip()

# Keep only needed columns
df = df[["organism group", "assembly", "AST phenotype"]].copy()

# Drop completely empty rows
df = df.dropna(subset=["AST phenotype"])

# Convert to string
df["AST phenotype"] = df["AST phenotype"].astype(str).str.strip()

# Remove useless UI text
df = df[df["AST phenotype"].str.lower() != "expand all"]

# Identify antibiotic names vs phenotype values
phenotype_pattern = r"^(Resistant|Susceptible|Intermediate)\s*\(\d+\)$"

current_antibiotic = None
rows = []

for _, row in df.iterrows():
    value = row["AST phenotype"]

    # If value is an antibiotic name
    if not re.match(phenotype_pattern, value, flags=re.IGNORECASE) and "(" not in value:
        current_antibiotic = value.lower().strip()
        continue

    # If value is a phenotype label
    if re.match(phenotype_pattern, value, flags=re.IGNORECASE):
        if current_antibiotic is not None:
            pheno = value.split("(")[0].strip().capitalize()
            rows.append({
                "assembly": row["assembly"],
                "organism group": row["organism group"],
                "antibiotic": current_antibiotic,
                "phenotype": pheno
            })

ast_long = pd.DataFrame(rows)

print("✅ Reconstructed long AST table shape:", ast_long.shape)
print(ast_long.head())

ast_long.to_csv("/content/ast_long_labels.csv", index=False)
print("📁 Saved: /content/ast_long_labels.csv")
```

```python
import pandas as pd
import re

# Load original
df = pd.read_excel("/content/original 1000 samples.xlsx")
df.columns = df.columns.str.strip()

# Keep only what we need
df = df[["organism group", "assembly", "AST phenotype"]].copy()

# Clean
df["AST phenotype"] = df["AST phenotype"].astype(str).str.strip()
df = df[df["AST phenotype"].str.lower() != "expand all"]

# Forward fill IDs because the file uses grouped layout
df["organism group"] = df["organism group"].ffill()
df["assembly"] = df["assembly"].ffill()

# Patterns
count_pat = re.compile(r"^(Resistant|Susceptible|Intermediate)\s*\((\d+)\)$", re.I)
other_pat = re.compile(r"^Other\s*\(\d+\)$", re.I)

# State variables
current_category = None  # 'Resistant'/'Susceptible'/'Intermediate'
current_counts = {"Resistant": None, "Intermediate": None, "Susceptible": None}
current_seen = {"Resistant": 0, "Intermediate": 0, "Susceptible": 0}

# Output rows
rows = []
qc_rows = []

# Track current genome key
prev_key = (None, None)

def flush_qc(key, counts, seen):
    """Store QC summary for a genome if we have any counts recorded."""
    asm, org = key
    if asm is None or org is None:
        return
    if any(v is not None for v in counts.values()):
        qc_rows.append({
            "assembly": asm,
            "organism group": org,
            "R_expected": counts["Resistant"],
            "I_expected": counts["Intermediate"],
            "S_expected": counts["Susceptible"],
            "R_seen": seen["Resistant"],
            "I_seen": seen["Intermediate"],
            "S_seen": seen["Susceptible"],
        })

for _, r in df.iterrows():
    asm = str(r["assembly"]).strip()
    org = str(r["organism group"]).strip()
    val = str(r["AST phenotype"]).strip()

    # Skip empty-like
    if val.lower() in ["nan", "none", ""]:
        continue
    if other_pat.match(val):
        continue

    key = (asm, org)

    # If we moved to a new genome block, flush QC and reset per-genome counters
    if prev_key != (None, None) and key != prev_key:
        flush_qc(prev_key, current_counts, current_seen)
        current_category = None
        current_counts = {"Resistant": None, "Intermediate": None, "Susceptible": None}
        current_seen = {"Resistant": 0, "Intermediate": 0, "Susceptible": 0}

    prev_key = key

    # If this is a category header like "Resistant (4)"
    m = count_pat.match(val)
    if m:
        cat = m.group(1).capitalize()
        cnt = int(m.group(2))
        current_category = cat
        current_counts[cat] = cnt
        # reset seen for this category when header appears
        current_seen[cat] = 0
        continue

    # Otherwise it's an antibiotic name line (belongs to current_category)
    # Only record if we are currently inside a category
    if current_category is not None:
        antibiotic = val.lower().strip()
        # Convert category -> phenotype label
        phenotype = {"Resistant": "R", "Intermediate": "I", "Susceptible": "S"}[current_category]

        rows.append({
            "assembly": asm,
            "organism group": org,
            "antibiotic": antibiotic,
            "phenotype": phenotype
        })
        current_seen[current_category] += 1

# Flush QC for the last genome
flush_qc(prev_key, current_counts, current_seen)

# Create long dataset
ast_long = pd.DataFrame(rows).drop_duplicates()

print("✅ Long AST dataset shape:", ast_long.shape)
print(ast_long.head(10))

ast_long.to_csv("/content/genome_antibiotic_RSI_long.csv", index=False)
print("📁 Saved: /content/genome_antibiotic_RSI_long.csv")

# QC table
qc = pd.DataFrame(qc_rows)
qc["R_ok"] = qc["R_expected"].fillna(0).astype(int) == qc["R_seen"]
qc["I_ok"] = qc["I_expected"].fillna(0).astype(int) == qc["I_seen"]
qc["S_ok"] = qc["S_expected"].fillna(0).astype(int) == qc["S_seen"]
qc["all_ok"] = qc["R_ok"] & qc["I_ok"] & qc["S_ok"]

print("\n✅ QC summary:")
print(qc["all_ok"].value_counts(dropna=False))

qc.to_csv("/content/genome_antibiotic_RSI_QC.csv", index=False)
print("📁 Saved: /content/genome_antibiotic_RSI_QC.csv")

# Wide dataset: one column per antibiotic with R/S/I
wide = ast_long.pivot_table(
    index=["assembly", "organism group"],
    columns="antibiotic",
    values="phenotype",
    aggfunc=lambda x: x.iloc[0]  # should be unique after cleaning
).reset_index()

wide.to_csv("/content/genome_antibiotic_RSI_wide.csv", index=False)
print("📁 Saved: /content/genome_antibiotic_RSI_wide.csv")
print("✅ Wide shape:", wide.shape)
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import GroupShuffleSplit
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report
from tensorflow.keras import layers, Model

# ----------------------------
# Load merged dataset
# ----------------------------
df = pd.read_csv("/content/final_AST_kmers_long.csv")

# ----------------------------
# Encoders
# ----------------------------
y_le = LabelEncoder()
org_le = LabelEncoder()
abx_le = LabelEncoder()

y = y_le.fit_transform(df["phenotype"].astype(str))  # R/S/I
org = org_le.fit_transform(df["organism group"].astype(str))
abx = abx_le.fit_transform(df["antibiotic"].astype(str))

num_classes = len(y_le.classes_)
num_org = len(org_le.classes_)
num_abx = len(abx_le.classes_)

print("✅ Phenotype classes:", y_le.classes_)

# ----------------------------
# K-mer features
# ----------------------------
drop_cols = ["assembly", "organism group", "antibiotic", "phenotype", "asm_key"]
kmer_cols = [c for c in df.columns if c not in drop_cols]

X = df[kmer_cols].astype(np.float32).values        # (N,4096)
X = X.reshape(-1, 256, 16)                         # (N,256,16)

# ----------------------------
# Leakage-free split by genome
# ----------------------------
groups = df["asm_key"].astype(str).values
gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups=groups))

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]
org_train, org_test = org[train_idx], org[test_idx]
abx_train, abx_test = abx[train_idx], abx[test_idx]

df_test = df.iloc[test_idx].copy()  # for per-antibiotic reporting

print("✅ Train:", X_train.shape, "Test:", X_test.shape)
print("✅ Train genomes:", len(set(groups[train_idx])), "Test genomes:", len(set(groups[test_idx])))

# ----------------------------
# Class weights (handles imbalance)
# ----------------------------
class_weights = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(y_train),
    y=y_train
)
class_weight_dict = {i: w for i, w in enumerate(class_weights)}
print("✅ Class weights:", class_weight_dict)

# ----------------------------
# Attention Layer
# ----------------------------
class AttentionLayer(layers.Layer):
    def __init__(self, units=64):
        super().__init__()
        self.W = layers.Dense(units, activation="tanh")
        self.V = layers.Dense(1)

    def call(self, inputs):
        score = self.V(self.W(inputs))           # (batch, time, 1)
        weights = tf.nn.softmax(score, axis=1)   # (batch, time, 1)
        context = tf.reduce_sum(weights * inputs, axis=1)
        return context

# ----------------------------
# Build CNN–LSTM–Attention model
# ----------------------------
seq_in = layers.Input(shape=(256, 16), name="kmer_seq")
org_in = layers.Input(shape=(1,), name="org_id")
abx_in = layers.Input(shape=(1,), name="abx_id")

x = layers.Conv1D(64, 3, activation="relu", padding="same")(seq_in)
x = layers.MaxPooling1D(2)(x)
x = layers.Dropout(0.2)(x)

x = layers.Conv1D(128, 3, activation="relu", padding="same")(x)
x = layers.MaxPooling1D(2)(x)
x = layers.Dropout(0.2)(x)

x = layers.Bidirectional(layers.LSTM(64, return_sequences=True))(x)
x = AttentionLayer(64)(x)

org_emb = layers.Embedding(num_org, 16)(org_in)
org_emb = layers.Flatten()(org_emb)

abx_emb = layers.Embedding(num_abx, 16)(abx_in)
abx_emb = layers.Flatten()(abx_emb)

x = layers.Concatenate()([x, org_emb, abx_emb])
x = layers.Dense(128, activation="relu")(x)
x = layers.Dropout(0.3)(x)
out = layers.Dense(num_classes, activation="softmax")(x)

model = Model([seq_in, org_in, abx_in], out)
model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-3),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# ----------------------------
# Train
# ----------------------------
callbacks = [
    tf.keras.callbacks.EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True)
]

history = model.fit(
    {"kmer_seq": X_train, "org_id": org_train, "abx_id": abx_train},
    y_train,
    validation_split=0.2,
    epochs=30,
    batch_size=32,
    callbacks=callbacks,
    class_weight=class_weight_dict
)

# ----------------------------
# Evaluate overall
# ----------------------------
pred = model.predict({"kmer_seq": X_test, "org_id": org_test, "abx_id": abx_test})
y_pred = np.argmax(pred, axis=1)

print("\n✅ Overall classification report:")
print(classification_report(y_test, y_pred, target_names=y_le.classes_))

# ----------------------------
# Evaluate per antibiotic
# ----------------------------
df_test = df_test.reset_index(drop=True)
df_test["y_true"] = y_test
df_test["y_pred"] = y_pred
df_test["y_true_lbl"] = y_le.inverse_transform(df_test["y_true"])
df_test["y_pred_lbl"] = y_le.inverse_transform(df_test["y_pred"])

reports = []
for abx_name, g in df_test.groupby("antibiotic"):
    if len(g) < 20:   # skip tiny groups
        continue
    rep = classification_report(
        g["y_true"], g["y_pred"],
        output_dict=True,
        zero_division=0
    )
    reports.append({
        "antibiotic": abx_name,
        "n": len(g),
        "accuracy": rep["accuracy"]
    })

abx_report = pd.DataFrame(reports).sort_values("accuracy", ascending=False)
print("\n✅ Per-antibiotic accuracy (top 15):")
print(abx_report.head(15))

abx_report.to_csv("/content/per_antibiotic_accuracy.csv", index=False)
print("📁 Saved: /content/per_antibiotic_accuracy.csv")

# ----------------------------
# Save model + encoders
# ----------------------------
model.save("/content/cnn_lstm_attention_AST_model.keras")
pd.Series(y_le.classes_).to_csv("/content/phenotype_classes.csv", index=False)
pd.Series(org_le.classes_).to_csv("/content/organism_classes.csv", index=False)
pd.Series(abx_le.classes_).to_csv("/content/antibiotic_classes.csv", index=False)

print("\n📁 Saved model + class mappings in /content/")
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import GroupShuffleSplit
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    classification_report, confusion_matrix,
    roc_auc_score, average_precision_score,
    balanced_accuracy_score, f1_score
)
from tensorflow.keras import layers, Model

# ----------------------------
# Load merged dataset
# ----------------------------
df = pd.read_csv("/content/final_AST_kmers_long.csv")

# Keep only R and S
df = df[df["phenotype"].isin(["R", "S"])].copy()
df["y_bin"] = (df["phenotype"] == "R").astype(int)  # R=1, S=0

print("✅ Binary dataset shape:", df.shape)
print(df["phenotype"].value_counts())

# ----------------------------
# Encode organism + antibiotic
# ----------------------------
org_le = LabelEncoder()
abx_le = LabelEncoder()

org = org_le.fit_transform(df["organism group"].astype(str))
abx = abx_le.fit_transform(df["antibiotic"].astype(str))

num_org = len(org_le.classes_)
num_abx = len(abx_le.classes_)

# ----------------------------
# K-mer features
# ----------------------------
drop_cols = ["assembly", "organism group", "antibiotic", "phenotype", "asm_key", "y_bin"]
kmer_cols = [c for c in df.columns if c not in drop_cols]

X = df[kmer_cols].astype(np.float32).values
X = X.reshape(-1, 256, 16)

y = df["y_bin"].values
groups = df["asm_key"].astype(str).values

# ----------------------------
# Leakage-free split by genome
# ----------------------------
gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups=groups))

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]
org_train, org_test = org[train_idx], org[test_idx]
abx_train, abx_test = abx[train_idx], abx[test_idx]

df_test = df.iloc[test_idx].copy().reset_index(drop=True)

print("✅ Train:", X_train.shape, "Test:", X_test.shape)
print("✅ Train genomes:", len(set(groups[train_idx])), "Test genomes:", len(set(groups[test_idx])))

# ----------------------------
# Class weights (binary)
# ----------------------------
cw = compute_class_weight(class_weight="balanced", classes=np.array([0,1]), y=y_train)
class_weight = {0: float(cw[0]), 1: float(cw[1])}
print("✅ Class weights:", class_weight)

# ----------------------------
# Attention Layer
# ----------------------------
class AttentionLayer(layers.Layer):
    def __init__(self, units=64):
        super().__init__()
        self.W = layers.Dense(units, activation="tanh")
        self.V = layers.Dense(1)

    def call(self, inputs):
        score = self.V(self.W(inputs))
        weights = tf.nn.softmax(score, axis=1)
        context = tf.reduce_sum(weights * inputs, axis=1)
        return context

# ----------------------------
# Model (binary output)
# ----------------------------
seq_in = layers.Input(shape=(256, 16), name="kmer_seq")
org_in = layers.Input(shape=(1,), name="org_id")
abx_in = layers.Input(shape=(1,), name="abx_id")

x = layers.Conv1D(64, 3, activation="relu", padding="same")(seq_in)
x = layers.MaxPooling1D(2)(x)
x = layers.Dropout(0.2)(x)

x = layers.Conv1D(128, 3, activation="relu", padding="same")(x)
x = layers.MaxPooling1D(2)(x)
x = layers.Dropout(0.2)(x)

x = layers.Bidirectional(layers.LSTM(64, return_sequences=True))(x)
x = AttentionLayer(64)(x)

org_emb = layers.Embedding(num_org, 16)(org_in)
org_emb = layers.Flatten()(org_emb)

abx_emb = layers.Embedding(num_abx, 16)(abx_in)
abx_emb = layers.Flatten()(abx_emb)

x = layers.Concatenate()([x, org_emb, abx_emb])
x = layers.Dense(128, activation="relu")(x)
x = layers.Dropout(0.3)(x)

# sigmoid for binary probability of Resistant
out = layers.Dense(1, activation="sigmoid")(x)

model = Model([seq_in, org_in, abx_in], out)
model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-3),
    loss="binary_crossentropy",
    metrics=[tf.keras.metrics.AUC(name="auc")]
)

model.summary()

# ----------------------------
# Train
# ----------------------------
callbacks = [
    tf.keras.callbacks.EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True)
]

history = model.fit(
    {"kmer_seq": X_train, "org_id": org_train, "abx_id": abx_train},
    y_train,
    validation_split=0.2,
    epochs=30,
    batch_size=32,
    callbacks=callbacks,
    class_weight=class_weight
)

# ----------------------------
# Evaluate
# ----------------------------
p = model.predict({"kmer_seq": X_test, "org_id": org_test, "abx_id": abx_test}).ravel()
y_pred = (p >= 0.5).astype(int)

print("\n✅ Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\n✅ Classification Report:\n", classification_report(y_test, y_pred, target_names=["S","R"]))

print("✅ Balanced accuracy:", balanced_accuracy_score(y_test, y_pred))
print("✅ F1:", f1_score(y_test, y_pred))
print("✅ AUROC:", roc_auc_score(y_test, p))
print("✅ AUPRC:", average_precision_score(y_test, p))

# ----------------------------
# Per-antibiotic AUROC (skip if only one class)
# ----------------------------
df_test["y_true"] = y_test
df_test["p_resistant"] = p

rows = []
for abx_name, g in df_test.groupby("antibiotic"):
    if len(g) < 30:
        continue
    if g["y_true"].nunique() < 2:
        continue
    rows.append({
        "antibiotic": abx_name,
        "n": len(g),
        "auroc": roc_auc_score(g["y_true"], g["p_resistant"]),
        "auprc": average_precision_score(g["y_true"], g["p_resistant"]),
    })

abx_perf = pd.DataFrame(rows).sort_values("auroc", ascending=False)
print("\n✅ Per-antibiotic AUROC (top 15):")
print(abx_perf.head(15))

abx_perf.to_csv("/content/per_antibiotic_binary_perf.csv", index=False)
print("📁 Saved: /content/per_antibiotic_binary_perf.csv")

# Save model + encoders
model.save("/content/cnn_lstm_attention_AST_binary.keras")
pd.Series(org_le.classes_).to_csv("/content/organism_classes.csv", index=False)
pd.Series(abx_le.classes_).to_csv("/content/antibiotic_classes.csv", index=False)

print("\n📁 Saved model: /content/cnn_lstm_attention_AST_binary.keras")
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import (
    precision_score, recall_score, f1_score,
    accuracy_score, balanced_accuracy_score,
    roc_auc_score, average_precision_score
)

# ----------------------------
# Inputs required:
# df_test must contain antibiotic column (from earlier)
# y_test = true binary labels (0=S, 1=R)
# p = predicted probabilities for Resistant
# ----------------------------

thr = 0.55
y_pred_thr = (p >= thr).astype(int)

# Create evaluation frame
eval_df = df_test.copy().reset_index(drop=True)
eval_df["y_true"] = y_test
eval_df["p_resistant"] = p
eval_df["y_pred"] = y_pred_thr

results = []

for abx_name, g in eval_df.groupby("antibiotic"):
    n = len(g)
    if n < 20:  # skip very small antibiotics
        continue

    y_t = g["y_true"].values
    y_h = g["y_pred"].values
    p_r = g["p_resistant"].values

    # Confusion matrix parts
    TP = int(((y_t == 1) & (y_h == 1)).sum())
    FP = int(((y_t == 0) & (y_h == 1)).sum())
    FN = int(((y_t == 1) & (y_h == 0)).sum())
    TN = int(((y_t == 0) & (y_h == 0)).sum())

    prec_r = precision_score(y_t, y_h, zero_division=0)
    rec_r  = recall_score(y_t, y_h, zero_division=0)
    f1_r   = f1_score(y_t, y_h, zero_division=0)

    acc = accuracy_score(y_t, y_h)
    bal_acc = balanced_accuracy_score(y_t, y_h)

    # AUROC/AUPRC only valid if both classes exist
    if len(np.unique(y_t)) == 2:
        auroc = roc_auc_score(y_t, p_r)
        auprc = average_precision_score(y_t, p_r)
    else:
        auroc = np.nan
        auprc = np.nan

    results.append({
        "antibiotic": abx_name,
        "n": n,
        "resistant_rate": float(np.mean(y_t)),
        "TP": TP, "FP": FP, "FN": FN, "TN": TN,
        "precision_R": prec_r,
        "recall_R": rec_r,
        "f1_R": f1_r,
        "accuracy": acc,
        "balanced_accuracy": bal_acc,
        "AUROC": auroc,
        "AUPRC": auprc
    })

per_abx = pd.DataFrame(results)

# Sort for paper (best F1_R or best AUROC)
per_abx_sorted = per_abx.sort_values(["f1_R", "AUROC"], ascending=False)

print("✅ Per-antibiotic table shape:", per_abx_sorted.shape)
print(per_abx_sorted.head(15))

# Save outputs
per_abx_sorted.to_csv("/content/per_antibiotic_binary_threshold055.csv", index=False)
print("📁 Saved: /content/per_antibiotic_binary_threshold055.csv")

# Also save a compact version (good for manuscript tables)
compact = per_abx_sorted[[
    "antibiotic", "n", "resistant_rate",
    "precision_R", "recall_R", "f1_R",
    "balanced_accuracy", "AUROC", "AUPRC"
]].copy()

compact.to_csv("/content/per_antibiotic_binary_threshold055_compact.csv", index=False)
print("📁 Saved: /content/per_antibiotic_binary_threshold055_compact.csv")
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import GroupShuffleSplit
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    confusion_matrix, classification_report,
    balanced_accuracy_score, f1_score,
    roc_auc_score, average_precision_score,
    precision_score, recall_score
)

from tensorflow.keras import layers, Model

# =========================================================
# 1) LOAD CLEAN DATA (NO CONFLICTS)
# =========================================================
df = pd.read_csv("/content/final_AST_kmers_long_no_conflicts.csv")

# Keep only R and S for binary
df = df[df["phenotype"].isin(["R", "S"])].copy()
df["y_bin"] = (df["phenotype"] == "R").astype(int)  # R=1, S=0

print("✅ Binary dataset shape:", df.shape)
print(df["phenotype"].value_counts())

# =========================================================
# 2) ENCODE ORGANISM + ANTIBIOTIC
# =========================================================
org_le = LabelEncoder()
abx_le = LabelEncoder()

org = org_le.fit_transform(df["organism group"].astype(str))
abx = abx_le.fit_transform(df["antibiotic"].astype(str))

num_org = len(org_le.classes_)
num_abx = len(abx_le.classes_)

# =========================================================
# 3) KMER FEATURES -> (256,16)
# =========================================================
drop_cols = ["assembly", "organism group", "antibiotic", "phenotype", "asm_key", "y_bin"]
kmer_cols = [c for c in df.columns if c not in drop_cols]

X = df[kmer_cols].astype(np.float32).values  # (N, 4096)
X = X.reshape(-1, 256, 16)

y = df["y_bin"].values
groups = df["asm_key"].astype(str).values

# =========================================================
# 4) LEAKAGE-FREE SPLIT BY GENOME
# =========================================================
gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups=groups))

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]
org_train, org_test = org[train_idx], org[test_idx]
abx_train, abx_test = abx[train_idx], abx[test_idx]

df_test = df.iloc[test_idx].copy().reset_index(drop=True)

print("✅ Train:", X_train.shape, "Test:", X_test.shape)
print("✅ Train genomes:", len(set(groups[train_idx])), "Test genomes:", len(set(groups[test_idx])))
print("✅ Genome overlap:", len(set(groups[train_idx]).intersection(set(groups[test_idx]))))

# =========================================================
# 5) CLASS WEIGHTS
# =========================================================
cw = compute_class_weight(class_weight="balanced", classes=np.array([0, 1]), y=y_train)
class_weight = {0: float(cw[0]), 1: float(cw[1])}
print("✅ Class weights:", class_weight)

# =========================================================
# 6) ATTENTION LAYER
# =========================================================
class AttentionLayer(layers.Layer):
    def __init__(self, units=64):
        super().__init__()
        self.W = layers.Dense(units, activation="tanh")
        self.V = layers.Dense(1)

    def call(self, inputs):
        score = self.V(self.W(inputs))
        weights = tf.nn.softmax(score, axis=1)
        context = tf.reduce_sum(weights * inputs, axis=1)
        return context

# =========================================================
# 7) BUILD CNN–BiLSTM–ATTENTION MODEL (BINARY)
# =========================================================
seq_in = layers.Input(shape=(256, 16), name="kmer_seq")
org_in = layers.Input(shape=(1,), name="org_id")
abx_in = layers.Input(shape=(1,), name="abx_id")

x = layers.Conv1D(64, 3, activation="relu", padding="same")(seq_in)
x = layers.MaxPooling1D(2)(x)
x = layers.Dropout(0.2)(x)

x = layers.Conv1D(128, 3, activation="relu", padding="same")(x)
x = layers.MaxPooling1D(2)(x)
x = layers.Dropout(0.2)(x)

x = layers.Bidirectional(layers.LSTM(64, return_sequences=True))(x)
x = AttentionLayer(64)(x)

org_emb = layers.Embedding(num_org, 16)(org_in)
org_emb = layers.Flatten()(org_emb)

abx_emb = layers.Embedding(num_abx, 16)(abx_in)
abx_emb = layers.Flatten()(abx_emb)

x = layers.Concatenate()([x, org_emb, abx_emb])
x = layers.Dense(128, activation="relu")(x)
x = layers.Dropout(0.3)(x)

out = layers.Dense(1, activation="sigmoid")(x)

model = Model([seq_in, org_in, abx_in], out)
model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-3),
    loss="binary_crossentropy",
    metrics=[tf.keras.metrics.AUC(name="auc")]
)

model.summary()

# =========================================================
# 8) TRAIN
# =========================================================
callbacks = [
    tf.keras.callbacks.EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True)
]

history = model.fit(
    {"kmer_seq": X_train, "org_id": org_train, "abx_id": abx_train},
    y_train,
    validation_split=0.2,
    epochs=30,
    batch_size=32,
    callbacks=callbacks,
    class_weight=class_weight
)

# =========================================================
# 9) EVALUATE (DEFAULT THRESHOLD 0.50)
# =========================================================
p = model.predict({"kmer_seq": X_test, "org_id": org_test, "abx_id": abx_test}).ravel()
y_pred_05 = (p >= 0.50).astype(int)

print("\n✅ Threshold = 0.50")
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_05))
print(classification_report(y_test, y_pred_05, target_names=["S", "R"]))
print("Balanced accuracy:", balanced_accuracy_score(y_test, y_pred_05))
print("F1(R):", f1_score(y_test, y_pred_05))
print("AUROC:", roc_auc_score(y_test, p))
print("AUPRC:", average_precision_score(y_test, p))

# =========================================================
# 10) FIND BEST THRESHOLD (MAX F1 FOR RESISTANT)
# =========================================================
thresholds = np.arange(0.05, 0.96, 0.05)
best = {"thr": None, "f1": -1, "precision": None, "recall": None}

for t in thresholds:
    y_hat = (p >= t).astype(int)
    f1 = f1_score(y_test, y_hat)
    prec = precision_score(y_test, y_hat, zero_division=0)
    rec = recall_score(y_test, y_hat)

    if f1 > best["f1"]:
        best = {"thr": float(t), "f1": float(f1), "precision": float(prec), "recall": float(rec)}

print("\n✅ Best threshold (max F1 for Resistant):", best)

# =========================================================
# 11) PER-ANTIBIOTIC PERFORMANCE AT BEST THRESHOLD
# =========================================================
thr = best["thr"]
y_pred_thr = (p >= thr).astype(int)

df_test = df_test.copy()
df_test["y_true"] = y_test
df_test["p_resistant"] = p
df_test["y_pred"] = y_pred_thr

rows = []
for abx_name, g in df_test.groupby("antibiotic"):
    if len(g) < 30:
        continue
    if g["y_true"].nunique() < 2:
        continue

    y_t = g["y_true"].values
    y_h = g["y_pred"].values
    p_r = g["p_resistant"].values

    rows.append({
        "antibiotic": abx_name,
        "n": len(g),
        "resistant_rate": float(np.mean(y_t)),
        "precision_R": float(precision_score(y_t, y_h, zero_division=0)),
        "recall_R": float(recall_score(y_t, y_h, zero_division=0)),
        "f1_R": float(f1_score(y_t, y_h, zero_division=0)),
        "balanced_accuracy": float(balanced_accuracy_score(y_t, y_h)),
        "AUROC": float(roc_auc_score(y_t, p_r)),
        "AUPRC": float(average_precision_score(y_t, p_r))
    })

per_abx = pd.DataFrame(rows).sort_values("f1_R", ascending=False)
print("\n✅ Per-antibiotic performance (top 15):")
print(per_abx.head(15))

per_abx.to_csv("/content/per_antibiotic_binary_bestthr_clean.csv", index=False)
print("📁 Saved: /content/per_antibiotic_binary_bestthr_clean.csv")

# =========================================================
# 12) SAVE MODEL + ENCODERS
# =========================================================
model.save("/content/cnn_lstm_attention_AST_binary_clean.keras")
pd.Series(org_le.classes_).to_csv("/content/organism_classes_clean.csv", index=False)
pd.Series(abx_le.classes_).to_csv("/content/antibiotic_classes_clean.csv", index=False)

print("\n📁 Saved model: /content/cnn_lstm_attention_AST_binary_clean.keras")
print("📁 Saved encoders: organism_classes_clean.csv, antibiotic_classes_clean.csv")
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.model_selection import GroupKFold
from sklearn.preprocessing import LabelEncoder
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    roc_auc_score, average_precision_score,
    balanced_accuracy_score, f1_score,
    precision_score, recall_score
)
from tensorflow.keras import layers, Model

# =========================
# Config
# =========================
DATA_PATH = "/content/final_AST_kmers_long_no_conflicts.csv"
N_SPLITS = 5
EPOCHS = 30
BATCH_SIZE = 32
RANDOM_SEED = 42

tf.random.set_seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

# =========================
# Load + Binary filter
# =========================
df = pd.read_csv(DATA_PATH)

# Binary (R/S only)
df = df[df["phenotype"].isin(["R", "S"])].copy()
df["y_bin"] = (df["phenotype"] == "R").astype(int)  # R=1, S=0

print("✅ Binary dataset:", df.shape)
print(df["phenotype"].value_counts())

# =========================
# Encode organism + antibiotic (fit on full dataset for stable ids)
# =========================
org_le = LabelEncoder()
abx_le = LabelEncoder()

df["org_id"] = org_le.fit_transform(df["organism group"].astype(str))
df["abx_id"] = abx_le.fit_transform(df["antibiotic"].astype(str))

num_org = len(org_le.classes_)
num_abx = len(abx_le.classes_)

# =========================
# Build X (k-mers)
# =========================
drop_cols = ["assembly", "organism group", "antibiotic", "phenotype", "asm_key", "y_bin", "org_id", "abx_id"]
kmer_cols = [c for c in df.columns if c not in drop_cols]

X_kmer = df[kmer_cols].astype(np.float32).values  # (N,4096)
X_seq = X_kmer.reshape(-1, 256, 16)               # (N,256,16)

y = df["y_bin"].values.astype(int)
org = df["org_id"].values.astype(int)
abx = df["abx_id"].values.astype(int)
groups = df["asm_key"].astype(str).values

# =========================
# Attention Layer
# =========================
class AttentionLayer(layers.Layer):
    def __init__(self, units=64):
        super().__init__()
        self.W = layers.Dense(units, activation="tanh")
        self.V = layers.Dense(1)

    def call(self, inputs):
        score = self.V(self.W(inputs))          # (B,T,1)
        weights = tf.nn.softmax(score, axis=1)  # (B,T,1)
        context = tf.reduce_sum(weights * inputs, axis=1)  # (B,H)
        return context

# =========================
# Model builder (fresh model each fold)
# =========================
def build_model(num_org, num_abx):
    seq_in = layers.Input(shape=(256, 16), name="kmer_seq")
    org_in = layers.Input(shape=(1,), name="org_id")
    abx_in = layers.Input(shape=(1,), name="abx_id")

    x = layers.Conv1D(64, 3, activation="relu", padding="same")(seq_in)
    x = layers.MaxPooling1D(2)(x)
    x = layers.Dropout(0.2)(x)

    x = layers.Conv1D(128, 3, activation="relu", padding="same")(x)
    x = layers.MaxPooling1D(2)(x)
    x = layers.Dropout(0.2)(x)

    x = layers.Bidirectional(layers.LSTM(64, return_sequences=True))(x)
    x = AttentionLayer(64)(x)

    org_emb = layers.Embedding(num_org, 16)(org_in)
    org_emb = layers.Flatten()(org_emb)

    abx_emb = layers.Embedding(num_abx, 16)(abx_in)
    abx_emb = layers.Flatten()(abx_emb)

    x = layers.Concatenate()([x, org_emb, abx_emb])
    x = layers.Dense(128, activation="relu")(x)
    x = layers.Dropout(0.3)(x)

    out = layers.Dense(1, activation="sigmoid")(x)

    model = Model([seq_in, org_in, abx_in], out)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(1e-3),
        loss="binary_crossentropy",
        metrics=[tf.keras.metrics.AUC(name="auc")]
    )
    return model

# =========================
# Helper: threshold tuning for max F1 on a set
# =========================
def best_threshold_max_f1(y_true, p_resistant, step=0.05):
    thresholds = np.arange(0.05, 0.96, step)
    best = {"thr": 0.5, "f1": -1.0, "precision": None, "recall": None}
    for t in thresholds:
        y_hat = (p_resistant >= t).astype(int)
        f1 = f1_score(y_true, y_hat, zero_division=0)
        if f1 > best["f1"]:
            best = {
                "thr": float(t),
                "f1": float(f1),
                "precision": float(precision_score(y_true, y_hat, zero_division=0)),
                "recall": float(recall_score(y_true, y_hat, zero_division=0))
            }
    return best

# =========================
# 5-Fold GroupKFold CV
# =========================
gkf = GroupKFold(n_splits=N_SPLITS)

fold_rows = []
all_pred_rows = []

for fold, (train_idx, test_idx) in enumerate(gkf.split(X_seq, y, groups=groups), start=1):
    # Split
    X_train, X_test = X_seq[train_idx], X_seq[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]
    org_train, org_test = org[train_idx], org[test_idx]
    abx_train, abx_test = abx[train_idx], abx[test_idx]

    groups_train = groups[train_idx]
    groups_test = groups[test_idx]
    overlap = len(set(groups_train).intersection(set(groups_test)))

    print(f"\n========== Fold {fold}/{N_SPLITS} ==========")
    print("Train samples:", len(train_idx), "Test samples:", len(test_idx))
    print("Train genomes:", len(set(groups_train)), "Test genomes:", len(set(groups_test)))
    print("Genome overlap:", overlap)

    # Class weights per fold
    cw = compute_class_weight(class_weight="balanced", classes=np.array([0,1]), y=y_train)
    class_weight = {0: float(cw[0]), 1: float(cw[1])}
    print("Class weights:", class_weight)

    # Build and train
    model = build_model(num_org=num_org, num_abx=num_abx)

    callbacks = [
        tf.keras.callbacks.EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True)
    ]

    history = model.fit(
        {"kmer_seq": X_train, "org_id": org_train, "abx_id": abx_train},
        y_train,
        validation_split=0.2,  # internal split within train fold
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        callbacks=callbacks,
        class_weight=class_weight,
        verbose=0
    )

    # Predict probabilities
    p_test = model.predict({"kmer_seq": X_test, "org_id": org_test, "abx_id": abx_test}, verbose=0).ravel()

    # Ranking metrics (threshold-free)
    auroc = roc_auc_score(y_test, p_test) if len(np.unique(y_test)) == 2 else np.nan
    auprc = average_precision_score(y_test, p_test) if len(np.unique(y_test)) == 2 else np.nan

    # Choose threshold using TRAIN-FOLD validation behavior:
    # We'll tune threshold on train-fold predictions from a held-out split = use validation_split indices indirectly is hard,
    # so we tune on the fold's TEST set? (Not allowed).
    # Instead: tune threshold on TRAIN fold via an explicit small validation split:
    # We'll do a simple deterministic split from the train fold for threshold tuning.

    # --- explicit threshold-tune split (within train fold) ---
    n_tr = len(train_idx)
    rng = np.random.default_rng(RANDOM_SEED + fold)
    perm = rng.permutation(n_tr)
    val_size = int(0.2 * n_tr)
    val_sub = perm[:val_size]
    tr_sub  = perm[val_size:]

    # (No retrain; we just use the already-trained model and score the val subset)
    p_val = model.predict(
        {"kmer_seq": X_train[val_sub], "org_id": org_train[val_sub], "abx_id": abx_train[val_sub]},
        verbose=0
    ).ravel()

    best = best_threshold_max_f1(y_train[val_sub], p_val, step=0.05)
    thr = best["thr"]

    yhat_test = (p_test >= thr).astype(int)

    # Threshold metrics
    bal_acc = balanced_accuracy_score(y_test, yhat_test)
    f1r = f1_score(y_test, yhat_test, zero_division=0)
    prec_r = precision_score(y_test, yhat_test, zero_division=0)
    rec_r = recall_score(y_test, yhat_test, zero_division=0)

    fold_rows.append({
        "fold": fold,
        "n_train": len(train_idx),
        "n_test": len(test_idx),
        "train_genomes": len(set(groups_train)),
        "test_genomes": len(set(groups_test)),
        "auroc": auroc,
        "auprc": auprc,
        "thr_maxF1_on_trainVal": thr,
        "precision_R": prec_r,
        "recall_R": rec_r,
        "f1_R": f1r,
        "balanced_accuracy": bal_acc
    })

    # Save per-row predictions for later analysis
    fold_pred = pd.DataFrame({
        "fold": fold,
        "asm_key": groups_test,
        "antibiotic_id": abx_test,
        "organism_id": org_test,
        "y_true": y_test,
        "p_resistant": p_test,
        "y_pred": yhat_test,
        "threshold_used": thr
    })
    all_pred_rows.append(fold_pred)

# =========================
# Summaries + Save
# =========================
cv_df = pd.DataFrame(fold_rows)
print("\n✅ Fold metrics:")
print(cv_df)

summary = cv_df[["auroc","auprc","precision_R","recall_R","f1_R","balanced_accuracy"]].agg(["mean","std"])
print("\n✅ CV mean ± std:")
print(summary)

cv_df.to_csv("/content/cv_fold_metrics.csv", index=False)
summary.to_csv("/content/cv_summary_mean_std.csv")
pd.concat(all_pred_rows, ignore_index=True).to_csv("/content/cv_all_predictions.csv", index=False)

print("\n📁 Saved:")
print(" - /content/cv_fold_metrics.csv")
print(" - /content/cv_summary_mean_std.csv")
print(" - /content/cv_all_predictions.csv")
```
